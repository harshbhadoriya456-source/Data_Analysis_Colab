```python
with open("student.csv", "w") as f:
    f.write("Name,Marks\n")
    f.write("Harsh,90\n")


with open("student.csv", "r") as f:
    data = f.read()

df = pd.read_csv("student.csv")

df.to_json("student.json")
print(pd.read_json("student.json"))

df.to_excel("student.xlsx")
print(pd.read_excel("student.xlsx"))

df.to_xml("student.xml")
print(pd.read_xml("student.xml"))
```

```python
print("Shape:")
print(df.shape)
print("\nColumns:")
print(df.columns)
print("\nInfo:")
print(df.info)
print("\nData Types:")
print(df.dtypes)
```
