```python
import json
df = pd.read_csv('students.csv')

df.to_json('students.json', orient='records')
with open('students.json', 'r') as file:
    json_data = json.loads(file.read())

df1 = pd.DataFrame(json_data)
json_new = df1.to_json(orient='records')



print(json_new)
```

```python
for chunk in pd.read_csv('students.csv', chunksize=1000):
  total = 0
count = 0
total += chunk['marks'].sum()
count += chunk['marks'].count()

print("Total Marks:", total)
print("Total Count:", count)
print("Average Marks:", total / count)
```
