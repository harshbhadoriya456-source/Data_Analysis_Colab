```python
import pandas as pd
from io import StringIO

data = """name,age,marks
harsh,19,85
aviral,20,90
adarsh,19,88"""

df = pd.read_csv(StringIO(data))
json_data = df.to_json(orient="records")

df2 = pd.read_json(StringIO(json_data), orient="records")


print(df.equals(df2))
```
